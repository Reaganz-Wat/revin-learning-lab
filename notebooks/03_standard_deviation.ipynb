# 03 — Standard Deviation

## From variance to standard deviation

Variance measures spread but its **units are squared** (e.g., score²).
Taking the square root returns us to the original units — this is the
**standard deviation** $\sigma$.

$$\sigma = \sqrt{\sigma^2} = \sqrt{\frac{1}{n}\sum_{i=1}^{n}(x_i - \bar{x})^2}$$

### Intuition

$\sigma$ answers: *On average, how far are data points from the mean?*

### The 68-95-99.7 rule (for normal distributions)

| Range | % of data |
|-------|-----------|
| $\bar{x} \pm 1\sigma$ | ~68% |
| $\bar{x} \pm 2\sigma$ | ~95% |
| $\bar{x} \pm 3\sigma$ | ~99.7% |

---

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from utils.ml_utils import calculate_mean, calculate_std
from utils.plotting import plot_histogram
```

```python
df     = pd.read_csv('../data/fundamentals_data.csv')
scores = df['student_score'].values

mu  = calculate_mean(scores)
std = calculate_std(scores)

print(f'Mean : {mu:.2f}')
print(f'Std  : {std:.2f}')
print(f'Verify with numpy: {np.std(scores):.2f}')
```

## Visualising mean ± std bands

```python
fig, ax = plt.subplots(figsize=(10, 4))

ax.hist(scores, bins=20, color='#4C72B0', edgecolor='white', alpha=0.8, label='Scores')
ax.axvline(mu,           color='red',    linestyle='-',  linewidth=2,   label=f'Mean = {mu:.1f}')
ax.axvline(mu - std,     color='orange', linestyle='--', linewidth=1.5, label=f'−1σ = {mu-std:.1f}')
ax.axvline(mu + std,     color='orange', linestyle='--', linewidth=1.5, label=f'+1σ = {mu+std:.1f}')
ax.axvline(mu - 2 * std, color='green',  linestyle=':',  linewidth=1.5, label=f'−2σ = {mu-2*std:.1f}')
ax.axvline(mu + 2 * std, color='green',  linestyle=':',  linewidth=1.5, label=f'+2σ = {mu+2*std:.1f}')

ax.axvspan(mu - std, mu + std, alpha=0.1, color='orange', label='68% region')
ax.axvspan(mu - 2*std, mu + 2*std, alpha=0.05, color='green', label='95% region')

ax.set_title('Student Scores — mean ± std bands', fontsize=13)
ax.set_xlabel('Score')
ax.set_ylabel('Frequency')
ax.legend(fontsize=8)
ax.spines[['top','right']].set_visible(False)
plt.tight_layout()
plt.show()
```

## Count values within each band

```python
n = len(scores)
in_1std = np.sum(np.abs(scores - mu) <= 1 * std)
in_2std = np.sum(np.abs(scores - mu) <= 2 * std)
in_3std = np.sum(np.abs(scores - mu) <= 3 * std)

print(f'Within ±1σ : {in_1std}/{n} = {100*in_1std/n:.1f}%  (theory: 68%)')
print(f'Within ±2σ : {in_2std}/{n} = {100*in_2std/n:.1f}%  (theory: 95%)')
print(f'Within ±3σ : {in_3std}/{n} = {100*in_3std/n:.1f}% (theory: 99.7%)')
```

## Std across all columns

```python
print(f"{'Column':<15} {'Mean':>8} {'Std':>8}")
print('-' * 33)
for col in df.columns:
    m = calculate_mean(df[col].values)
    s = calculate_std(df[col].values)
    print(f'{col:<15} {m:>8.2f} {s:>8.2f}')
```

Notice `sales` has a much larger std than `student_score`. This is why we need
normalization — to put all features on a comparable scale.
